# 01. Import Libraries

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import os
```

# 02. Creating Directory for reports and visuals

```python
os.makedirs("../../visuals_and_reports/modeling_analysis", exist_ok=True)
```

# 03. Loading experimental results

```python
with open("../../models/experiment_results.json", "r") as f:
    results = json.load(f)
```

# 04. Creating Dataframe from Jason for better overview

```python
df = pd.DataFrame(results).T
df = df.drop(["best_model", "best_f1_score"], errors="ignore")
```

```python
df
```

# 05. Plotting Accuracy, F1-Score and ROC_AUC for individual models

```python
df[["accuracy", "f1_score", "roc_auc"]].plot(kind="bar", figsize=(10,6))
plt.title("Model Comparison")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../../visuals_and_reports/modeling_analysis/model_comparison.png")
plt.show()
```

# 06. Key Observation
* Overall Gradient_boosting outperforms in all matrices
