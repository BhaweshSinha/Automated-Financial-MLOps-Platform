# 01. Import Libraries

```python
import os
import pandas as pd
import numpy as np
```

# 02. Load Data

```python
data_path = "../../data/raw_data"
datasets = {}
for file in os.listdir(data_path):
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(data_path, file))
        df["Ticker"] = file.replace(".csv", "")
        datasets[file] = df
print(f"Total datasets loaded: {len(datasets)}")
```

# 03. Combined Datasets Summary

```python
summary_list = []
for name, df in datasets.items():
    info = {
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Start_Date": df["Date"].min() if "Date" in df.columns else None,
        "End_Date": df["Date"].max() if "Date" in df.columns else None,
        "Columns_Names": list(df.columns)
    }
    summary_list.append(info)
summary_df = pd.DataFrame(summary_list)
summary_df
```

# 04. Combined Datasets Price Summary

```python
market_summary = []
for name, df in datasets.items():
    stats = {
        "Dataset": name,
        "Mean_Close": df["Close"].mean() if "Close" in df.columns else None,
        "Std_Close": df["Close"].std() if "Close" in df.columns else None,
        "Min_Close": df["Close"].min() if "Close" in df.columns else None,
        "Max_Close": df["Close"].max() if "Close" in df.columns else None,
        "Avg_Volume": df["Volume"].mean() if "Volume" in df.columns else None
    }
    market_summary.append(stats)
market_summary_df = pd.DataFrame(market_summary)
market_summary_df
```

# 05. Combined Datasets Movements Summary

```python
movement_summary = []
for name, df in datasets.items():
    if "Close" in df.columns:
        df_sorted = df.sort_values("Date")
        df_sorted["Next_Close"] = df_sorted["Close"].shift(-1)
        df_sorted["Movement"] = np.where(
            df_sorted["Next_Close"] > df_sorted["Close"], "UP", "DOWN"
        )
        up_ratio = (df_sorted["Movement"] == "UP").mean()
        down_ratio = (df_sorted["Movement"] == "DOWN").mean()
        movement_summary.append({
            "Dataset": name,
            "UP_Ratio": up_ratio,
            "DOWN_Ratio": down_ratio
        })
movement_df = pd.DataFrame(movement_summary)
movement_df
```

# 06. Combined Datasets Volatility Summary

```python
volatility_summary = []
for name, df in datasets.items():
    if "Close" in df.columns:
        returns = df["Close"].pct_change()
        volatility = returns.std()

        volatility_summary.append({
            "Dataset": name,
            "Volatility": volatility
        })

volatility_df = pd.DataFrame(volatility_summary)
volatility_df.sort_values(by="Volatility", ascending=False)
```

# 07. Final Datasets merged Summary

```python
final_summary = summary_df.merge(market_summary_df, on="Dataset", how="left")
final_summary = final_summary.merge(movement_df, on="Dataset", how="left")
final_summary = final_summary.merge(volatility_df, on="Dataset", how="left")
final_summary
```
