# 01. Import Libraries

```python
import os
import pandas as pd
```

# 02. Load Data

```python
data_path = "../../data/raw_data"
datasets = []
for file in os.listdir(data_path):
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(data_path, file))
        df["Ticker"] = file.replace(".csv", "")
        datasets.append(df)

print(f"Total datasets loaded: {len(datasets)}")
```

# 03. Combined Datasets Shape

```python
combined_df = pd.concat(datasets, ignore_index=True)

print("Combined shape:", combined_df.shape)
```

# 04. Setting the Output Directory Path

```python
output_dir = "../../data/combined_dataset"
os.makedirs(output_dir, exist_ok=True)
print("Output directory ready:", output_dir)
```

# 05. Saving the Merged Dataset to the Directory

```python
output_path = os.path.join(output_dir, "combined_data.csv")
combined_df.to_csv(output_path, index=False)
print(f"File saved at: {output_path}")
```
