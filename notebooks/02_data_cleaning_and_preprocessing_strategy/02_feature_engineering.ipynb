# 01. Import Libraries

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

# 02. Load cleaned data

```python
df = pd.read_csv("../../data/interim/cleaned_data.csv")
print("Loaded Shape:", df.shape)
```

# 03. Feature Engineering

## 3.1 Returns

```python
df['Returns'] = df['Close'].pct_change()
```

## 3.2 Moving Averages

```python
df['Ma_7'] = df['Close'].rolling(7).mean()
df['Ma_30'] = df['Close'].rolling(30).mean()
```

## 3.3 Volatility

```python
df['Volatility'] = df['Returns'].rolling(7).std()
```

## 3.4 Lag Features

```python
df['Lag_1'] = df['Close'].shift(1)
```

# 04. Removing NaN Values

```python
df = df.dropna()
```

# 05. Final Checks

```python
print("After Feature Engineering Shape:", df.shape)
print("Missing Values:\n", df.isnull().sum())
```

# 06. Save Featured Data

```python
df.to_csv("../../data/interim/featured_data.csv", index=False)
```
