# 01. Import Libraries

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import joblib
import os
import warnings
warnings.filterwarnings('ignore')
```

# 02. Load feature-engineered data

```python
df = pd.read_csv("../../data/interim/featured_data.csv")
print("Loaded Shape:", df.shape)
```

# 03. Select features to scale

```python
features = ['Close', 'Returns', 'Ma_7', 'Ma_30', 'Volatility']
```

# 04. Scaling

```python
scaler = StandardScaler()
df[features] = scaler.fit_transform(df[features])
```

# 05. Validation

```python
assert df.isnull().sum().sum() == 0, "Missing values found!"
```

# 06. Save final processed data

```python
df.to_csv("../../data/interim/final_data.csv", index=False)
```
